Iphone sales analysis

```python
import pandas as pd
import numpy as numpy
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("apple_products.csv")
```

To see data

```python
df
```

Check for missing values

```python
print(df.isnull().sum())
```

Descriptive analysis It is shows only the numerical value

```python
print(df.describe())
```

Top 10 selling Iphone in india . According to rating

```python
highest_rated=df.sort_values(by=["Star Rating"], ascending=False)
highest_rated = highest_rated.head(10)
print(highest_rated['Product Name'])
```

Lets have a look at the number of rating of the highest rated i phone on flipkart

```python
import plotly.express as px

fig = px.bar(
    highest_rated.head(10),
    x="Product Name",
    y="Number Of Ratings",
    title="Top 10 Highest Rated Apple Products"
    ,color='Product Name'
)

fig.update_layout(
    width=1400,
    height=750,
    title_x=0.5,
    xaxis_title="Products",
    yaxis_title="Number Of Ratings ",
    font=dict(size=10),
    xaxis_tickangle=-45
)
fig.show()

```

Lets have a look at the number of  Star rating of the highest rated i phone on flipkart.

```python
figure = go.Figure(data=[go.Pie(labels=df['Product Name'], values=df['Number Of Ratings'], hole=.3)])
figure.show()
```

Relationship between sales price and Rating

```python
figure = px.scatter(df,  x ="Number Of Ratings", y ="Sale Price", size="Discount Percentage", trendline="ols", title="Relationship between Sales Price and Number of Ratings",color="Star Rating")
figure.show()
```

```python
figure = px.scatter(df,x = "Number Of Ratings",y="Discount Percentage",size ="Sale Price",trendline="ols",title="Relationship between Discount Percentage and Number of Ratings")
figure.show()
```
